La celda siguiente ejecuta sus celdas de código. Regenera salidas H3; use una copia para conservar evidencia.

```python
from pathlib import Path
import json
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "repositorio/configs/proyecto.json").is_file() or (p / "configs/proyecto.json").is_file())
if not (root / "configs/proyecto.json").is_file(): root = root / "repositorio"
notebook = json.loads((root / "notebooks/replicacion_calidad_datos.ipynb").read_text(encoding="utf-8"))
for cell in notebook["cells"]:
    if cell["cell_type"] == "code": exec(compile("".join(cell["source"]), "notebooks/iniciar_replicacion.ipynb", "exec"), globals())
```
